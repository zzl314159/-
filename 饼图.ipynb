```python
from random import randint

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.style as pls
from pandas import DataFrame

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
pls.use('ggplot')
plt.figure(figsize=(9,6))
#导入数据
df=pd.DataFrame(data={'区域':['华东','华北','东北','西北','西南','华南'],'销售数':np.random.randint(low=100,high=5000,size=6)})
x=df['销售数'].tolist()
labels=df['区域'].tolist()
explode=[0.05,0.05,0,0,0,0]
#画图
plt.pie(x,autopct='%1.1f%%',shadow=False,labels=labels,explode=explode,startangle=90,pctdistance=0.8,textprops={'fontsize':12,'color':'k'},counterclock=False)
plt.title('双十一各区域销售数占比')
plt.show()
```
