```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.style as pls

%matplotlib inline
%config InlineBackend.figure_format = 'svg'
plt.rcParams[('font.sans-serif')] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
pls.use("ggplot")
plt.figure(figsize=(9,6))
#导入数据
df=pd.DataFrame(data={'商品种类':['床品件套','厨房电器','汽车配件','浴室工具','卧室家具','电脑硬件','家装饰品','办公家具'],
'销售数':np.random.randint(low=100,high=5000,size=8)})
x=df['商品种类'].tolist()
y=df['销售数'].tolist()
#画条形图
plt.barh(x,width=y,height=0.5,align='center',label='销售数')
plt.title('不同商品种类销量',loc='center')
for a,b in zip(x,y):plt.text(b+100,a,b,ha='center',va='center',fontsize=10)
plt.xlabel('销售数')
plt.ylabel('商品种类')
plt.legend()
plt.show()
```
